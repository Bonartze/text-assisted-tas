```python
!pip install -q -U huggingface_hub

from huggingface_hub import list_repo_files
from pathlib import Path
from collections import Counter

repo_id = "CVML-TueAI/Breakfast-Actions"

files = list_repo_files(
    repo_id=repo_id,
    repo_type="dataset",
)

print("Total files in repo:", len(files))

print("\nFirst 300 files:")
for f in files[:300]:
    print(f)

print("\nSuffix counts:")
suffix_counts = Counter(Path(f).suffix.lower() for f in files)
for suffix, count in suffix_counts.most_common(30):
    print(repr(suffix), count)

print("\nFiles containing video/videos:")
for f in files:
    if "video" in f.lower():
        print(f)
```

```python
from pathlib import Path

video_suffixes = {".avi", ".mp4", ".mkv", ".webm", ".mov", ".mpg", ".mpeg", ".tar"}

video_files = [
    f for f in files
    if Path(f).suffix.lower() in video_suffixes
]

jsonl_files = [
    f for f in files
    if f.endswith(".jsonl")
]

print("video-like files:", len(video_files))
for f in video_files[:100]:
    print(f)

print("\njsonl files:", len(jsonl_files))
for f in jsonl_files[:50]:
    print(f)
```

```python
from huggingface_hub import snapshot_download
from pathlib import Path

repo_id = "CVML-TueAI/Breakfast-Actions"
local_dir = "/content/drive/MyDrive/mmf_tas_lab_data/breakfast_raw_videos"

allow = video_files + jsonl_files

print("Files to download:", len(allow))
print("Destination:", local_dir)

snapshot_download(
    repo_id=repo_id,
    repo_type="dataset",
    local_dir=local_dir,
    allow_patterns=allow,
    resume_download=True,
)

print("Downloaded to:", local_dir)
```

```python
from pathlib import Path

root = Path("/content/drive/MyDrive/mmf_tas_lab_data/breakfast_raw_videos")

videos = []
for ext in ["*.avi", "*.mp4", "*.mkv", "*.webm", "*.mov", "*.mpg", "*.mpeg"]:
    videos.extend(root.rglob(ext))

videos = sorted(set(videos))

print("num videos:", len(videos))
for p in videos[:50]:
    print(p.relative_to(root))
```
